```python
# 8O — Create normalized position families

def normalize_position_family(pos):
    if pd.isna(pos):
        return "Unknown"

    # Split club-level labels, then split each label into position tags.
    positions = set()

    for label in str(pos).split("|"):
        positions.update(
            tag.strip()
            for tag in label.split(",")
            if tag.strip()
        )

    # Goalkeepers are kept separate.
    if "GK" in positions:
        return "GK"

    # Use a consistent order regardless of the original label order.
    position_order = ["DF", "MF", "FW"]
    positions = [
        position for position in position_order
        if position in positions
    ]

    return "/".join(positions) if positions else "Unknown"


df_model_pool["Position_Family"] = (
    df_model_pool["Pos"].apply(normalize_position_family)
)

print("Normalized position-family counts:")
display(
    df_model_pool["Position_Family"]
    .value_counts(dropna=False)
    .to_frame(name="Player count")
)

print("\nExamples of normalized multi-position players:")
display(
    df_model_pool.loc[
        df_model_pool["Pos"].str.contains(r"\|", na=False),
        ["Player", "Pos", "Position_Family", "Min"]
    ].head(15)
)

print(
    "\nMissing normalized position families:",
    df_model_pool["Position_Family"].isna().sum()
)
```

```python

# Cell 8W: First version of the player recommendation function

def recommend_similar_players(player_name, top_n=5):
    # Find the player (case-insensitive exact match)
    matches = np.flatnonzero(
        df_features["Player"].str.casefold().eq(
            player_name.strip().casefold()
        ).to_numpy()
    )

    if len(matches) == 0:
        raise ValueError(f"Player not found: {player_name}")

    if len(matches) > 1:
        raise ValueError(
            "Multiple matching player records found. "
            "We need to resolve player identity first."
        )

    player_idx = matches[0]
    player_position = df_features.iloc[player_idx]["Position_Family"]

    # Split hybrid positions into their constituent roles
    player_roles = set(player_position.split("/"))


    # Find candidates with at least one compatible role
    candidate_roles = df_features["Position_Family"].apply(
        lambda pos: bool(player_roles & set(pos.split("/")))
    ).to_numpy().copy()

    # Exclude the selected player
    candidate_roles[player_idx] = False

    candidate_indices = np.flatnonzero(candidate_roles)

    
    # Exclude the selected player
    candidate_roles[player_idx] = False

    candidate_indices = np.flatnonzero(candidate_roles)

    # Retrieve similarity scores for eligible candidates
    scores = similarity_matrix[player_idx, candidate_indices]

    # Rank candidates from most to least similar
    ranked_order = np.argsort(scores)[::-1][:top_n]
    best_indices = candidate_indices[ranked_order]
    best_scores = scores[ranked_order]

    # Build readable results
    result_columns = [
        col for col in ["Player", "Position_Family", "Age", "Min"]
        if col in df_features.columns
    ]

    recommendations = df_features.iloc[best_indices][
        result_columns
    ].copy()

    recommendations.insert(
        0,
        "Similarity_Score",
        np.round(best_scores, 4)
    )

    return recommendations.reset_index(drop=True)


# First test: Martin Ødegaard
recommend_similar_players("Martin Ødegaard", top_n=5)
```

```python
def get_similar_players(
    player_name,
    similarity_matrix,
    top_n=5
):
    player_indices = df_outfield.index[
        df_outfield["Player"].str.casefold()
        == player_name.casefold()
    ].tolist()

    if not player_indices:
        print(f"Player not found: {player_name}")
        return None

    player_index = df_outfield.index.get_loc(
        player_indices[0]
    )

    similarity_scores = similarity_matrix[player_index]

    similar_indices = np.argsort(similarity_scores)[::-1]

    similar_indices = [
        idx for idx in similar_indices
        if idx != player_index
    ]

    top_indices = similar_indices[:top_n]

    recommendations = df_outfield.iloc[top_indices][
        ["Player", "Position_Family", "Age", "Squad"]
    ].copy()

    recommendations["Similarity"] = [
        similarity_scores[idx] for idx in top_indices
    ]

    return recommendations.reset_index(drop=True)
```

```python

def recommend_from_matrix(player_name, matrix, top_n=5):
    # Use the same exact, case-insensitive player lookup
    matches = np.flatnonzero(
        df_features["Player"].str.casefold().eq(
            player_name.strip().casefold()
        ).to_numpy()
    )

    if len(matches) == 0:
        raise ValueError(f"Player not found: {player_name}")

    if len(matches) > 1:
        raise ValueError(
            f"Multiple matching records for {player_name}"
        )

    player_idx = matches[0]
    player_position = df_features.iloc[player_idx][
        "Position_Family"
    ]
    player_roles = set(player_position.split("/"))

    # Apply the same compatible-position filter
    candidate_roles = df_features["Position_Family"].apply(
        lambda pos: bool(
            player_roles & set(pos.split("/"))
        )
    ).to_numpy().copy()

    # Exclude the selected player
    candidate_roles[player_idx] = False
    candidate_indices = np.flatnonzero(candidate_roles)

    # Rank using the supplied model's similarity matrix
    scores = matrix[player_idx, candidate_indices]
    ranked_order = np.argsort(scores)[::-1][:top_n]
    best_indices = candidate_indices[ranked_order]
    best_scores = scores[ranked_order]

    result_columns = [
        col for col in [
            "Player", "Position_Family", "Age", "Min"
        ]
        if col in df_features.columns
    ]

    recommendations = df_features.iloc[
        best_indices
    ][result_columns].copy()

    recommendations.insert(
        0, "Similarity_Score", np.round(best_scores, 4)
    )

    return recommendations.reset_index(drop=True)


# Compare the original and candidate models
for player in ["Martin Ødegaard", "Kevin De Bruyne"]:
    print(f"\n{'=' * 65}")
    print(f"QUERY PLAYER: {player}")
    print("=" * 65)

    print("\nBASELINE — 21 features")
    display(
        recommend_from_matrix(
            player, similarity_matrix, top_n=5
        )
    )

    print("\nCANDIDATE — 20 features")
    display(
        recommend_from_matrix(
            player, candidate_similarity_matrix, top_n=5
        )
    )
```

```python
def normalize_position_family(pos):
    if pd.isna(pos):
        return "Unknown"

    positions = set()

    for label in str(pos).split("|"):
        positions.update(
            tag.strip()
            for tag in label.split(",")
            if tag.strip()
        )

    if "GK" in positions:
        return "GK"

    position_order = ["DF", "MF", "FW"]

    positions = [
        position for position in position_order
        if position in positions
    ]

    return "/".join(positions) if positions else "Unknown"


df_model_pool["Position_Family"] = (
    df_model_pool["Pos"].apply(normalize_position_family)
)

print("Position distribution:")
display(df_model_pool["Position_Family"].value_counts())

print(
    "\nMissing position families:",
    df_model_pool["Position_Family"].isna().sum()
)
```

```python
def get_similar_players(
    player_name,
    similarity_matrix,
    top_n=5
):
    player_indices = df_outfield.index[
        df_outfield["Player"].str.casefold()
        == player_name.casefold()
    ].tolist()

    if not player_indices:
        print(f"Player not found: {player_name}")
        return None

    player_index = df_outfield.index.get_loc(
        player_indices[0]
    )

    similarity_scores = similarity_matrix[player_index]

    similar_indices = np.argsort(similarity_scores)[::-1]

    similar_indices = [
        idx for idx in similar_indices
        if idx != player_index
    ]

    top_indices = similar_indices[:top_n]

    recommendations = df_outfield.iloc[top_indices][
        ["Player", "Position_Family", "Age", "Squad"]
    ].copy()

    recommendations["Similarity"] = [
        similarity_scores[idx] for idx in top_indices
    ]

    return recommendations.reset_index(drop=True)
```

```python
def normalize_position_family(pos):
    if pd.isna(pos):
        return "Unknown"

    positions = set()

    for label in str(pos).split("|"):
        positions.update(
            tag.strip()
            for tag in label.split(",")
            if tag.strip()
        )

    if "GK" in positions:
        return "GK"

    position_order = ["DF", "MF", "FW"]

    positions = [
        position
        for position in position_order
        if position in positions
    ]

    return "/".join(positions) if positions else "Unknown"


df_model_pool["Position_Family"] = (
    df_model_pool["Pos"].apply(normalize_position_family)
)

print("Position distribution:")
display(df_model_pool["Position_Family"].value_counts())

print("\nMissing position values:")
print(df_model_pool["Position_Family"].isna().sum())
```

```python
df_features = df_model_pool.copy()

playing_time_90s = df_features["Min"] / 90

per90_features = [
    "Gls", "Ast", "xG", "xAG", "npxG",
    "Sh", "SoT",
    "Tkl", "TklW", "Int", "Blocks", "Clr",
    "Touches", "Carries", "PrgC", "PrgP", "PrgR",
    "KP", "xA", "PPA", "Recov"
]

for feature in per90_features:
    df_features[f"{feature}_per90"] = (
        df_features[feature] / playing_time_90s
    )

print("Feature dataset shape:", df_features.shape)
print("Number of per-90 features:", len(per90_features))

per90_columns = [f"{feature}_per90" for feature in per90_features]

print(
    "Missing per-90 values:",
    df_features[per90_columns].isna().sum().sum()
)
```
