```python
# Inspect all players who appear more than once

player_counts = df_2024["Player"].value_counts()

repeated_players = player_counts[player_counts > 1].index

duplicate_rows = (
    df_2024[df_2024["Player"].isin(repeated_players)]
    .sort_values("Player")
)

duplicate_rows[
    ["Player", "Pos", "Squad", "Comp", "Age", "Min"]
].head(50)
```

```python
# Compare repeated-player records by club and league

duplicate_summary = (
    duplicate_rows
    .groupby("Player")
    .agg(
        number_of_rows=("Player", "size"),
        number_of_teams=("Squad", "nunique"),
        number_of_leagues=("Comp", "nunique"),
        positions=("Pos", lambda x: ", ".join(sorted(set(x)))),
        total_minutes=("Min", "sum")
    )
    .sort_values("number_of_rows", ascending=False)
)

duplicate_summary.head(20)
```

```python
# Check whether duplicate player rows are completely identical

duplicate_groups = (
    df_2024[
        df_2024["Player"].isin(repeated_players)
    ]
    .groupby("Player")
)

identical_duplicate_players = []

for player, group in duplicate_groups:
    if group.drop_duplicates().shape[0] == 1:
        identical_duplicate_players.append(player)

print(
    "Players with completely identical duplicate rows:",
    len(identical_duplicate_players)
)

print(identical_duplicate_players[:20])
```

```python
# 9. Position family audit

def get_position_family(pos):
    if pd.isna(pos):
        return "Unknown"

    positions = set(pos.split(","))

    if "GK" in positions:
        return "GK"
    elif "DF" in positions and "MF" in positions:
        return "DF/MF"
    elif "MF" in positions and "FW" in positions:
        return "MF/FW"
    elif "DF" in positions and "FW" in positions:
        return "DF/FW"
    elif "DF" in positions:
        return "DF"
    elif "MF" in positions:
        return "MF"
    elif "FW" in positions:
        return "FW"
    else:
        return "Unknown"


position_audit = df_2024[["Player", "Pos"]].copy()

position_audit["Position_Family"] = position_audit["Pos"].apply(
    get_position_family
)

print("Position family distribution:")
print(position_audit["Position_Family"].value_counts())

print("\nPosition family percentages:")
print(
    (position_audit["Position_Family"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# 8E — Minutes coverage by position family

position_minutes_summary = (
    df_2024.groupby("Position_Family")["Min"]
    .agg(
        players="count",
        mean_minutes="mean",
        median_minutes="median",
        low_minutes_450=("Min", lambda x: (x < 450).sum()),
        low_minutes_900=("Min", lambda x: (x < 900).sum()),
        min_minutes="min",
        max_minutes="max"
    )
)

position_minutes_summary["low_450_percent"] = (
    position_minutes_summary["low_minutes_450"]
    / position_minutes_summary["players"] * 100
).round(2)

position_minutes_summary["low_900_percent"] = (
    position_minutes_summary["low_minutes_900"]
    / position_minutes_summary["players"] * 100
).round(2)

position_minutes_summary.sort_values("players", ascending=False)
```
