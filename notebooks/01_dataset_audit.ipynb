```python
import pandas as pd
from pathlib import Path

# Project root directory
PROJECT_ROOT = Path.cwd().parent

# Dataset paths
DATA_DIR = PROJECT_ROOT / "data" / "raw"

file_2024 = DATA_DIR / "players_data-2024_2025.csv"
file_2025 = DATA_DIR / "players_data-2025_2026 (1).csv"

# Load datasets
df_2024 = pd.read_csv(file_2024)
df_2025 = pd.read_csv(file_2025)

print("2024/25 dataset loaded successfully.")
print("2025/26 dataset loaded successfully.")
```

```python
print("2024/25 shape:", df_2024.shape)
print("2025/26 shape:", df_2025.shape)
```

```python
columns_2024 = set(df_2024.columns)
columns_2025 = set(df_2025.columns)

print("Columns only in 2024/25:")
print(sorted(columns_2024 - columns_2025))

print("\nColumns only in 2025/26:")
print(sorted(columns_2025 - columns_2024))

print("\nNumber of columns in 2024/25:", len(columns_2024))
print("Number of columns in 2025/26:", len(columns_2025))
print("Same columns:", columns_2024 == columns_2025)
```

```python
common_columns = sorted(columns_2024 & columns_2025)

print("Number of common columns:", len(common_columns))
print("\nCommon columns:")
for column in common_columns:
    print(column)
```

```python
pd.set_option("display.max_rows", 150)

column_info = pd.DataFrame({
    "column": df_2025.columns,
    "dtype": df_2025.dtypes.astype(str),
    "missing": df_2025.isna().sum(),
    "unique_values": df_2025.nunique()
})

column_info
```
